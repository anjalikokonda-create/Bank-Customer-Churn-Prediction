```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('/content/Churn_Modelling.csv')
```

```python
geo_churn = df.groupby('Geography')['Exited'].mean() * 100

geo_churn.plot(kind='bar')
plt.title('Geography-wise Churn Rate')
plt.xlabel('Country')
plt.ylabel('Churn Rate (%)')
plt.show()
```

```python
gender_churn = df.groupby('Gender')['Exited'].mean() * 100

gender_churn.plot(kind='bar')
plt.title('Gender-wise Churn Rate')
plt.xlabel('Gender')
plt.ylabel('Churn Rate (%)')
plt.show()
```

```python
df['AgeGroup'] = pd.cut(
    df['Age'],
    bins=[0,30,45,60,100],
    labels=['<30','30-45','46-60','60+']
)
```

```python
age_churn = df.groupby('AgeGroup')['Exited'].mean() * 100

age_churn.plot(kind='bar')
plt.title('Age Group-wise Churn')
plt.xlabel('Age Group')
plt.ylabel('Churn Rate (%)')
plt.show()
```

```python
active_churn = df.groupby('IsActiveMember')['Exited'].mean() * 100

active_churn.plot(kind='bar')
plt.title('Active vs Inactive Customer Churn')
plt.xlabel('IsActiveMember')
plt.ylabel('Churn Rate (%)')
plt.show()
```

```python
df['BalanceSegment'] = pd.cut(
    df['Balance'],
    bins=[-1,0,100000,df['Balance'].max()],
    labels=['Zero Balance','Low Balance','High Balance']
)
```

```python
balance_churn = df.groupby('BalanceSegment')['Exited'].mean() * 100

balance_churn.plot(kind='bar')
plt.title('Balance Segment Churn')
plt.xlabel('Balance Segment')
plt.ylabel('Churn Rate (%)')
plt.show()
```

```python
df['Exited'].value_counts().plot(
    kind='pie',
    autopct='%1.1f%%'
)

plt.title('Overall Customer Churn')
plt.ylabel('')
plt.show()
```

```python
plt.figure(figsize=(6,4))

sns.histplot(df['CreditScore'], bins=20)

plt.title('Credit Score Distribution')
plt.xlabel('Credit Score')
plt.ylabel('Count')
plt.show()
```

```python
plt.figure(figsize=(6,4))

sns.boxplot(x='Exited', y='Balance', data=df)

plt.title('Balance vs Churn')
plt.show()
```

```python
plt.figure(figsize=(6,4))

sns.boxplot(x='Exited', y='EstimatedSalary', data=df)

plt.title('Salary vs Churn')
plt.show()
```

```python
tenure_churn = df.groupby('Tenure')['Exited'].mean() * 100

tenure_churn.plot(kind='line', marker='o', figsize=(6,4))

plt.title('Tenure-wise Churn')
plt.xlabel('Tenure')
plt.ylabel('Churn Rate (%)')
plt.show()
```

```python
plt.figure(figsize=(10,8))

sns.heatmap(df.corr(numeric_only=True), annot=True)

plt.title('Correlation Heatmap')
plt.show()
```
