```python
# ==============================
# IMPORT LIBRARIES
# ==============================
import pandas as pd
import matplotlib.pyplot as plt

# Treemap safe import
try:
    import squarify
    treemap_available = True
except:
    treemap_available = False


# ==============================
# LOAD DATA
# ==============================
data = pd.read_csv("Churn_Modelling.csv")

# Create label
data['Churn Status'] = data['Exited'].map({0: 'Retained', 1: 'Churned'})

# Age groups
data['Age Group'] = pd.cut(
    data['Age'],
    bins=[18, 30, 50, 100],
    labels=['Young', 'Middle Age', 'Senior']
)

# Balance groups
data['Balance Group'] = pd.cut(
    data['Balance'],
    bins=[-1, 50000, 100000, 300000],
    labels=['Low', 'Medium', 'High']
)


# ==============================
# CREATE DASHBOARD
# ==============================
plt.figure(figsize=(18,12))


# ------------------------------
# 1. DONUT CHART
# ------------------------------
plt.subplot(3,3,1)

churn_counts = data['Churn Status'].value_counts()

plt.pie(churn_counts.values,
        labels=churn_counts.index,
        autopct='%1.1f%%',
        colors=['#2ecc71', '#e74c3c'])

centre_circle = plt.Circle((0,0), 0.70, fc='white')
plt.gca().add_artist(centre_circle)

plt.title("Churn Distribution")


# ------------------------------
# 2. BAR: GEOGRAPHY
# ------------------------------
plt.subplot(3,3,2)

pd.crosstab(data['Geography'], data['Churn Status']).plot(
    kind='bar', ax=plt.gca())

plt.title("Geography vs Churn")
plt.xticks(rotation=0)


# ------------------------------
# 3. BAR: GENDER
# ------------------------------
plt.subplot(3,3,3)

pd.crosstab(data['Gender'], data['Churn Status']).plot(
    kind='bar', ax=plt.gca())

plt.title("Gender vs Churn")
plt.xticks(rotation=0)


# ------------------------------
# 4. BAR: AGE GROUP
# ------------------------------
plt.subplot(3,3,4)

pd.crosstab(data['Age Group'], data['Churn Status']).plot(
    kind='bar', ax=plt.gca())

plt.title("Age Group vs Churn")
plt.xticks(rotation=0)


# ------------------------------
# 5. BAR: BALANCE
# ------------------------------
plt.subplot(3,3,5)

pd.crosstab(data['Balance Group'], data['Churn Status']).plot(
    kind='bar', ax=plt.gca())

plt.title("Balance vs Churn")
plt.xticks(rotation=0)


# ------------------------------
# 6. LINE: AGE vs BALANCE (GENDER)
# ------------------------------
plt.subplot(3,3,6)

grouped = data.groupby(['Age', 'Gender'])['Balance'].mean().unstack()

for gender in grouped.columns:
    plt.plot(grouped.index, grouped[gender], label=gender)

plt.title("Balance vs Age (Gender)")
plt.legend()
plt.grid()


# ------------------------------
# 7. SCATTER: AGE vs BALANCE
# ------------------------------
plt.subplot(3,3,7)

colors = data['Exited'].map({0:'green', 1:'red'})

plt.scatter(data['Age'], data['Balance'], c=colors, alpha=0.5)

plt.title("Age vs Balance")
plt.xlabel("Age")
plt.ylabel("Balance")


# ------------------------------
# 8. SCATTER: CREDIT vs BALANCE
# ------------------------------
plt.subplot(3,3,8)

plt.scatter(data['CreditScore'], data['Balance'], c=colors, alpha=0.5)

plt.title("Credit Score vs Balance")
plt.xlabel("Credit Score")


# ------------------------------
# 9. TREEMAP
# ------------------------------
plt.subplot(3,3,9)

if treemap_available:
    squarify.plot(
        sizes=churn_counts.values,
        label=churn_counts.index,
        color=['#2ecc71', '#e74c3c'],
        alpha=0.8
    )
    plt.axis('off')
    plt.title("Treemap")
else:
    plt.text(0.3,0.5,"Install squarify", fontsize=12)


# ==============================
# FINAL DISPLAY
# ==============================
plt.tight_layout()
plt.show()
```
